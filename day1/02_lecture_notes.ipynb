```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
bench_1k = %timeit -o -r 30 -n 50 np.random.random(1_000)
df_1k = pd.DataFrame()
df_1k['times'] = bench_1k.timings
df_1k['level'] = 1_000
df_1k['method'] = 'np.random.random()'
df_1k

bench_10k = %timeit -o -r 30 -n 50 np.random.random(10_000)
df_10k = pd.DataFrame()
df_10k['times'] = bench_1k.timings
df_10k['level'] = 10_000
df_10k['method'] = 'np.random.random()'
df_10k

bench_100k = %timeit -o -r 30 -n 50 np.random.random(100_000)
df_100k = pd.DataFrame()
df_100k['times'] = bench_100k.timings
df_100k['level'] = 100_000
df_100k['method'] = 'np.random.random()'
df_100k

df  = pd.concat([df_1k, df_10k, df_100k], ignore_index=True)
df
```

```python
import matplotlib.pyplot as plt
```

```python

sns.lineplot(data=df, x='level', y='times', hue='method', estimator='min', linestyle='dotted', marker='o');
# plt.gca().margins(x=0.5, y=0.5)
plt.loglog();
# plt.xlim()
# plt.gca().get_xlim()
```
