```python
import numpy as np
```

```python
bench1 = %timeit -r 1000 -n 50 -o np.random.random(1000)
bench2 = %timeit -r 1000 -n 50 -o np.random.random(10_000)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.eventplot([bench1.timings, bench2.timings], alpha=.1);
```
